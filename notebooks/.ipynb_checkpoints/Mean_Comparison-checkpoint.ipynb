```python
from datetime import datetime

def datetime_to_seconds(datetime_str):
    dt = datetime.fromisoformat(datetime_str)
    epoch = datetime.fromisoformat("1970-01-01T00:00:00")
    # If your system timezone is not UTC, you may need to make both datetimes timezone-aware
    if dt.tzinfo is None:
        dt = dt.replace(tzinfo=None)  # or set to UTC: dt = dt.replace(tzinfo=timezone.utc)
    if epoch.tzinfo is None:
        epoch = epoch.replace(tzinfo=None)  # or set to UTC: epoch = epoch.replace(tzinfo=timezone.utc)
    # Calculate the difference in seconds
    return (dt - epoch).total_seconds()

def extract_machine(filename):
    # Match gl followed by any characters except dots, or ss2
    pattern = r'(gl[^._-]+|ss2)'
    match = re.search(pattern, filename)
    if match:
        return match.group(0)
```

# SAR

```python
# SAR runs are stored in '../runs/run.dpp-local-nc16*/cpu_utilization.csv' 
sar_dirs = glob.glob('../runs/run.dpp-local-nc16*/cpu_utilization.csv') # all CPU usage files
```

```python
# Generate a table containing the entry id, average CPU utilization
col_names=['Machine', 'Duration(m)','Avg CPU Usage(%)', 'Max CPU Usage(%)']
# sar_data = pd.DataFrame(columns=col_names)
rows = []

for x in sar_dirs:
    # skips the iteration if the file is empty
    try:
        df = pd.read_csv(x) # iterates the files 
    except pd.errors.EmptyDataError:
        continue
 
    all_data = df[df['CPU'] == 'all'] # get aggregates stats for all the cores
    # calculate the duration of a run
    timestamps = list(map(datetime_to_seconds, all_data['Timestamp'].to_list())) # converts dates in seconds
    duration = (timestamps[-1] - timestamps[0]) / 60
    # calculate CPU usage
    cpu_usage = all_data['%user'] + all_data['%system'] # list of CPU usage
    avg_cpu_usage = round(cpu_usage.mean(), 3)
    # get machine name
    machine = extract_machine(x)

    row = {
        col_names[0] : machine, col_names[1] : duration, col_names[2] : avg_cpu_usage, col_names[3] : cpu_usage.max()
    }
 
    rows.append(row)

pd.DataFrame(rows)
```

# SLURM

```python
def time_str_to_seconds(time_str):
    # Initialize days to 0
    days = 0
    
    # Check if day part exists
    if '-' in time_str:
        days_part, time_part = time_str.split('-', 1)
        days = int(days_part)
    else:
        time_part = time_str
    
    # Split time part by colon
    parts = time_part.split(':')
    
    # Initialize hours, minutes, seconds, milliseconds
    hours = 0
    minutes = 0
    seconds = 0
    milliseconds = 0
    
    # Parse based on number of parts
    if len(parts) == 3:
        # Format: HH:MM:SS.mmm or HH:MM:SS
        hours = int(parts[0])
        minutes = int(parts[1])
        sec_part = parts[2]
    elif len(parts) == 2:
        # Format: MM:SS.mmm or MM:SS
        minutes = int(parts[0])
        sec_part = parts[1]
    elif len(parts) == 1:
        # Format: SS.mmm or SS
        sec_part = parts[0]
    else:
        raise ValueError(f"Invalid time format: {time_str}")
    
    # Handle seconds and milliseconds
    if '.' in sec_part:
        sec_str, ms_str = sec_part.split('.', 1)
        seconds = int(sec_str)
        # Convert milliseconds string to int (pad or truncate to 3 digits)
        ms_str = (ms_str + "000")[:3]
        milliseconds = int(ms_str)
    else:
        seconds = int(sec_part)
    
    # Calculate total seconds including milliseconds as fractional seconds
    total_seconds = (
        days * 24 * 3600 +
        hours * 3600 +
        minutes * 60 +
        seconds +
        milliseconds / 1000.0
    )
    
    return total_seconds 
```
