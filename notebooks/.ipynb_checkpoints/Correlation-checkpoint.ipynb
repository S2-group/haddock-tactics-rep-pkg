```python
import glob
import pandas as pd
```

```python
dpp = pd.read_csv('../data/data_dpp.csv')
dpp_data = dpp[['Elapsed', 'ConsumedEnergy', 'NCPUS']]
```

```python
correlation = dpp_data['Elapsed'].corr(dpp_data['ConsumedEnergy'])
print(f"The Pearson correlation between Time and Energy is: {correlation:.2f}")
```
