```python
import parse_slurm as ps
import data_analysis as da
import glob
import pandas as pd
```

```python
def get_runs(data, scenarios = ['dpp', 'daa']):
    # number of runs for NCPUS
    runs = []
    for w in scenarios:
        for x in [4, 8, 16, 32]:
            number_runs = len(ps.get_group(data, workflow=w, ncpus=str(x)))
            val = {'number': int(number_runs), 'workflow': w, 'cpus': str(x)}
            runs.append(val)
    return pd.DataFrame(runs)

def get_runs_workflow(data):
    # number of runs for NCPUS
    runs = []
    for x in [4, 8, 16, 32]:
        number_runs = len(ps.get_cpus(data, ncpus=x))
        val = {'number': int(number_runs), 'cpus': x}
        runs.append(val)
    return pd.DataFrame(runs)

def get_runs_node(data):
    # number of runs for NCPUS
    runs = []
    for x in ['gl5', 'gl6', 'superserver2']:
        number_runs = len(ps.get_node(data, node=x))
        val = {'number': int(number_runs), 'node': x}
        runs.append(val)
    return pd.DataFrame(runs)
```
