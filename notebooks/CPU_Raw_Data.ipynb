```python
def time_str_to_seconds(time_str):
    # Initialize days to 0
    days = 0
    
    # Check if day part exists
    if '-' in time_str:
        days_part, time_part = time_str.split('-', 1)
        days = int(days_part)
    else:
        time_part = time_str
    
    # Split time part by colon
    parts = time_part.split(':')
    
    # Initialize hours, minutes, seconds, milliseconds
    hours = 0
    minutes = 0
    seconds = 0
    milliseconds = 0
    
    # Parse based on number of parts
    if len(parts) == 3:
        # Format: HH:MM:SS.mmm or HH:MM:SS
        hours = int(parts[0])
        minutes = int(parts[1])
        sec_part = parts[2]
    elif len(parts) == 2:
        # Format: MM:SS.mmm or MM:SS
        minutes = int(parts[0])
        sec_part = parts[1]
    elif len(parts) == 1:
        # Format: SS.mmm or SS
        sec_part = parts[0]
    else:
        raise ValueError(f"Invalid time format: {time_str}")
    
    # Handle seconds and milliseconds
    if '.' in sec_part:
        sec_str, ms_str = sec_part.split('.', 1)
        seconds = int(sec_str)
        # Convert milliseconds string to int (pad or truncate to 3 digits)
        ms_str = (ms_str + "000")[:3]
        milliseconds = int(ms_str)
    else:
        seconds = int(sec_part)
    
    # Calculate total seconds including milliseconds as fractional seconds
    total_seconds = (
        days * 24 * 3600 +
        hours * 3600 +
        minutes * 60 +
        seconds +
        milliseconds / 1000.0
    )
    
    return total_seconds
```
